```python
import pandas as pd
df = pd.read_csv('/content/e commerce dataset (1).csv', encoding='latin-1')
print(df.describe())
```

```python
print(df)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

print("Data types after conversion:")
print(df[['Order Date', 'Ship Date']].dtypes)
```

With the dates converted, we can now check for missing values or explore unique categories to understand the data better.

```python
print("Missing values per column:")
print(df.isnull().sum())
```

This output will show the count of missing values for each column. If there are any, we'll need to decide how to handle them (e.g., fill them, drop rows/columns).

```python
print("Unique values in 'Category':")
print(df['Category'].unique())
print("\nValue counts for 'Category':")
print(df['Category'].value_counts())

print("\nUnique values in 'Sub-Category':")
print(df['Sub-Category'].unique())
print("\nValue counts for 'Sub-Category':")
print(df['Sub-Category'].value_counts())
```

This will show us all the distinct categories and sub-categories present in the e-commerce data, along with how often each appears. This can be very useful for understanding the product catalog and for potential filtering or aggregation later on.

```python
total_sales_by_category = df.groupby('Category')['Sales'].sum()
print("Total Sales by Category:")
print(total_sales_by_category)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(x=total_sales_by_category.index, y=total_sales_by_category.values, palette='viridis')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# --- Selecting Records (Rows) based on a condition ---
# Example 1: Select rows where 'Sales' is greater than 500
print("\nRecords where Sales > 500 (first 5):")
display(df[df['Sales'] > 500].head())

# Example 2: Select rows where 'Category' is 'Technology'
print("\nRecords where Category is 'Technology' (first 5):")
display(df[df['Category'] == 'Technology'].head())

# Example 3: Select rows with multiple conditions (e.g., Sales > 500 AND Category is 'Technology')
print("\nRecords where Sales > 500 AND Category is 'Technology' (first 5):")
display(df[(df['Sales'] > 500) & (df['Category'] == 'Technology')].head())


# --- Deleting Records (Rows) based on a condition ---
# Instead of 'deleting' in place, it's common to create a new DataFrame that excludes certain rows.
# Example 1: Create a new DataFrame excluding rows where 'Profit' is negative
# This creates a new DataFrame 'df_positive_profit' containing only rows where Profit is non-negative
print("\nOriginal DataFrame shape:", df.shape)
df_positive_profit = df[df['Profit'] >= 0]
print("DataFrame shape after excluding negative profit records:", df_positive_profit.shape)


# --- Selecting Specific Columns ---
# Example 1: Select a single column
print("\n'Product Name' column (first 5):")
display(df['Product Name'].head())

# Example 2: Select multiple columns
print("\n'Category', 'Sub-Category', and 'Sales' columns (first 5):")
display(df[['Category', 'Sub-Category', 'Sales']].head())


# --- Deleting Columns ---
# Example 1: Delete a single column (creates a new DataFrame without the specified column)
# Let's say we want to remove 'Row ID'.
print("\nOriginal DataFrame columns:", df.columns.tolist()[:5], "...")
df_without_row_id = df.drop(columns=['Row ID'])
print("New DataFrame columns after dropping 'Row ID':", df_without_row_id.columns.tolist()[:5], "...")

# Example 2: Delete multiple columns
# Let's remove 'Customer ID' and 'Postal Code'.
print("\nOriginal DataFrame columns:", df.columns.tolist()[:5], "...")
df_reduced = df.drop(columns=['Customer ID', 'Postal Code'])
print("New DataFrame columns after dropping 'Customer ID' and 'Postal Code':", df_reduced.columns.tolist()[:5], "...")
```

```python
# --- Sorting Operations ---
# Example 1: Sort the DataFrame by 'Sales' in ascending order
print("\nDataFrame sorted by 'Sales' (top 5 rows, ascending):")
display(df.sort_values(by='Sales').head())

# Example 2: Sort the DataFrame by 'Profit' in descending order
print("\nDataFrame sorted by 'Profit' (top 5 rows, descending):")
display(df.sort_values(by='Profit', ascending=False).head())

# --- Ranking Operations ---
# Example 1: Rank 'Sales' in ascending order (default method is 'average')
print("\nRanking of 'Sales' (first 10 values):")
display(df['Sales'].rank().head(10))

# Example 2: Rank 'Profit' in descending order, using 'dense' method
print("\nRanking of 'Profit' (first 10 values, descending, dense rank):")
display(df['Profit'].rank(ascending=False, method='dense').head(10))
```

```python
print("\n--- Descriptive Statistics for Numerical Columns ---")
display(df.describe())

print("\n--- Specific Statistical Operations ---")
# Mean of 'Sales'
print(f"Average Sales: {df['Sales'].mean():.2f}")

# Median of 'Profit'
print(f"Median Profit: {df['Profit'].median():.2f}")

# Standard Deviation of 'Quantity'
print(f"Standard Deviation of Quantity: {df['Quantity'].std():.2f}")

# Minimum and Maximum of 'Sales'
print(f"Minimum Sales: {df['Sales'].min():.2f}")
print(f"Maximum Sales: {df['Sales'].max():.2f}")

# Mode of 'Ship Mode'
print("Mode of Ship Mode:")
display(df['Ship Mode'].mode())
```

```python
print("\n--- Uniqueness and Counts for Categorical Columns ---")

categorical_cols = ['Ship Mode', 'Segment', 'Category', 'Sub-Category']

for col in categorical_cols:
    print(f"\n--- Column: {col} ---")
    print(f"Unique values ({col}):")
    display(df[col].unique())
    print(f"Number of unique values ({col}): {df[col].nunique()}")
    print(f"Value counts ({col}):")
    display(df[col].value_counts())
```

```python
df_renamed = df.copy()

print("\n--- Renaming Columns ---")

# Example 1: Rename a single column
# Let's rename 'Row ID' to 'Record ID'
print("Original columns (before single rename):", df_renamed.columns.tolist()[:3], "...")
df_renamed_single = df_renamed.rename(columns={'Row ID': 'Record ID'})
print("Columns after renaming 'Row ID' to 'Record ID':", df_renamed_single.columns.tolist()[:3], "...")


# Example 2: Rename multiple columns
# Let's rename 'Order ID' to 'Transaction ID' and 'Customer Name' to 'Client Name'
print("\nOriginal columns (before multiple rename):", df_renamed.columns.tolist()[:5], "...")
df_renamed_multiple = df_renamed.rename(columns={
    'Order ID': 'Transaction ID',
    'Customer Name': 'Client Name'
})
print("Columns after renaming 'Order ID' and 'Customer Name':", df_renamed_multiple.columns.tolist()[:5], "...")

print("\nDisplaying head of DataFrame with multiple renamed columns:")
display(df_renamed_multiple.head())
```
