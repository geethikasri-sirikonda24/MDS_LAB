```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =======================================================
#                   Dataset Loading and Setup
# =======================================================

# Load the dataset
try:
    df = pd.read_csv('/content/NY-House-Dataset.csv')
    print("Dataset loaded successfully.")
    print("\nDataset Head:")
    print(df.head())
    print("\nDataset Columns:")
    print(df.columns.tolist())
    print("\nDataset Info:")
    df.info()
except FileNotFoundError:
    print("Error: The file '/content/NY-House-Dataset.csv' was not found.")
    print("Please ensure the dataset is uploaded to the correct path.")
    df = None # Set df to None to prevent further errors

if df is not None:
    numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
    categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
    print("\n--- Visualizations Starting Below ---")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# =======================================================
#                        Line Plot
# =======================================================

if df is not None:
    print("\nGenerating Line Plot...")
    # A line plot typically shows a trend over an ordered variable (like time or an index).
    # If you have specific time-series data or an ordered numerical feature, replace 'df.index' and the y-columns.
    # Example: sns.lineplot(x='Date_Column', y='PRICE', data=df)

    plt.figure(figsize=(12, 6))
    if 'PRICE' in df.columns and 'SQUARE_FEET' in df.columns: # Assuming common column names
        sns.lineplot(x=df.index, y='PRICE', data=df, label='PRICE')
        sns.lineplot(x=df.index, y='SQUARE_FEET', data=df, label='SQUARE_FEET')
        plt.title('Line Plot of Price and Square Feet over Index')
        plt.ylabel('Value')
    elif len(numerical_cols) >= 2:
        sns.lineplot(x=df.index, y=numerical_cols[0], data=df, label=numerical_cols[0])
        sns.lineplot(x=df.index, y=numerical_cols[1], data=df, label=numerical_cols[1])
        plt.title(f'Line Plot of {numerical_cols[0]} and {numerical_cols[1]} over Index')
        plt.ylabel('Value')
    elif len(numerical_cols) == 1:
        sns.lineplot(x=df.index, y=numerical_cols[0], data=df)
        plt.title(f'Line Plot of {numerical_cols[0]} over Index')
        plt.ylabel(numerical_cols[0])
    else:
        print("Not enough numerical columns for a meaningful Line Plot.")
    plt.xlabel('Data Index')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("Cannot perform Line Plot as the dataset could not be loaded.")
```

```python
# =======================================================
#                         Bar Plot
# =======================================================

if df is not None:
    print("\nGenerating Bar Plot...")
    # Bar plots are great for comparing categorical data. Adjust 'categorical_column' and 'numerical_column' as needed.
    # Example: sns.barplot(x='BOROUGH', y='PRICE', data=df)

    bar_x_col = None
    bar_y_col = None

    if 'BROKERTITLE' in df.columns and 'PRICE' in df.columns: # Using BROKERTITLE for variety
        bar_x_col = 'BROKERTITLE'
        bar_y_col = 'PRICE'
    elif len(categorical_cols) > 0 and len(numerical_cols) > 0:
        bar_x_col = categorical_cols[0]
        bar_y_col = numerical_cols[0]

    if bar_x_col and bar_y_col:
        # Increased figure height to accommodate long labels
        plt.figure(figsize=(12, 10))
        # Calculate mean for bar plot if multiple entries per category
        plot_data = df.groupby(bar_x_col)[bar_y_col].mean().reset_index()
        sns.barplot(x=bar_x_col, y=bar_y_col, data=plot_data)
        plt.title(f'Average {bar_y_col} by {bar_x_col}')
        plt.xlabel(bar_x_col)
        plt.ylabel(f'Average {bar_y_col}')
        plt.xticks(rotation=90, ha='right') # Rotate labels for better readability
        plt.grid(axis='y', linestyle='--')
        plt.tight_layout() # This should now work better with increased figure size
        plt.show()
    else:
        print("Not enough suitable categorical and numerical columns for Bar Plot.")
else:
    print("Cannot perform Bar Plot as the dataset could not be loaded.")
```

```python
# =======================================================
#                       Histogram
# =======================================================

if df is not None:
    print("\nGenerating Histogram...")
    # Histograms show the distribution of a single numerical variable. Adjust 'numerical_column' as needed.
    # Example: sns.histplot(df['PRICE'], kde=True)

    hist_col = None
    if 'PRICE' in df.columns:
        hist_col = 'PRICE'
    elif len(numerical_cols) > 0:
        hist_col = numerical_cols[0]

    if hist_col:
        plt.figure(figsize=(10, 6))
        sns.histplot(df[hist_col], bins=30, kde=True, color='skyblue')
        plt.title(f'Histogram of {hist_col} Distribution')
        plt.xlabel(hist_col)
        plt.ylabel('Frequency')
        plt.grid(axis='y', linestyle='--')
        plt.tight_layout()
        plt.show()
    else:
        print("No suitable numerical column found for Histogram.")
else:
    print("Cannot perform Histogram as the dataset could not be loaded.")
```

```python
# =======================================================
#                      Density Plot
# =======================================================

if df is not None:
    print("\nGenerating Density Plot...")
    # Density plots (KDE plots) are another way to visualize the distribution of a numerical variable.
    # Adjust 'numerical_column' as needed.
    # Example: sns.kdeplot(df['PRICE'], fill=True)

    if 'PRICE' in df.columns: # Using PRICE again for consistency or first numerical col
        kde_col = 'PRICE'
    elif len(numerical_cols) > 0:
        kde_col = numerical_cols[0]
    else:
        kde_col = None

    if kde_col:
        plt.figure(figsize=(10, 6))
        sns.kdeplot(df[kde_col], fill=True, color='salmon')
        plt.title(f'Density Plot of {kde_col} Distribution')
        plt.xlabel(kde_col)
        plt.ylabel('Density')
        plt.grid(True)
        plt.tight_layout()
        plt.show()
    else:
        print("No suitable numerical column found for Density Plot.")
else:
    print("Cannot perform Density Plot as the dataset could not be loaded.")
```

```python
# =======================================================
#                       Scatter Plot
# =======================================================

if df is not None:
    print("\nGenerating Scatter Plot...")
    # Scatter plots show the relationship between two numerical variables. Adjust 'x_column' and 'y_column' as needed.
    # Example: sns.scatterplot(x='SQUARE_FEET', y='PRICE', data=df)

    scatter_x_col = None
    scatter_y_col = None

    if 'SQUARE_FEET' in df.columns and 'PRICE' in df.columns:
        scatter_x_col = 'SQUARE_FEET'
        scatter_y_col = 'PRICE'
    elif len(numerical_cols) >= 2:
        scatter_x_col = numerical_cols[0]
        scatter_y_col = numerical_cols[1]

    if scatter_x_col and scatter_y_col:
        plt.figure(figsize=(10, 6))
        sns.scatterplot(x=scatter_x_col, y=scatter_y_col, data=df, alpha=0.7)
        plt.title(f'Scatter Plot of {scatter_y_col} vs {scatter_x_col}')
        plt.xlabel(scatter_x_col)
        plt.ylabel(scatter_y_col)
        plt.grid(True)
        plt.tight_layout()
        plt.show()
    else:
        print("Not enough suitable numerical columns for Scatter Plot.")
else:
    print("Cannot perform Scatter Plot as the dataset could not be loaded.")
```
