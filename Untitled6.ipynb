```python
import pandas as pd

# Load the CSV file into a pandas DataFrame
df = pd.read_csv('/content/HousePrices_HalfMil.csv.zip', compression='zip')

# Display the first 5 rows and information about the dataframe
display(df.head())
display(df.info())
```

```python
import statsmodels.api as sm

# Define the independent variable (X) and dependent variable (y)
X = df['Area']
y = df['Prices']

# Add a constant to the independent variable for the intercept term
X = sm.add_constant(X)

# Create and fit the simple linear regression model
model = sm.OLS(y, X)
results = model.fit()

# Calculate the residuals
residuals = results.resid

# Display the model summary and the first few residuals
print(results.summary())
print('\nFirst 5 residuals:')
print(residuals.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Residuals vs. Fitted values
sns.scatterplot(x=results.fittedvalues, y=residuals, ax=axes[0])
axes[0].axhline(y=0, color='r', linestyle='--')
axes[0].set_xlabel('Fitted Values')
axes[0].set_ylabel('Residuals')
axes[0].set_title('Residuals vs. Fitted Values')

# Plot 2: Histogram of residuals
sns.histplot(residuals, kde=True, ax=axes[1])
axes[1].set_xlabel('Residuals')
axes[1].set_ylabel('Frequency')
axes[1].set_title('Histogram of Residuals')

plt.tight_layout()
plt.show()

# Plot 3: Q-Q plot of residuals
plt.figure(figsize=(8, 6))
stats.probplot(residuals, dist="norm", plot=plt)
plt.title('Q-Q Plot of Residuals')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('Standardized Residuals')
plt.grid(True)
plt.show()
```

```python
from statsmodels.stats.diagnostic import linear_reset

# Perform the Ramsey RESET test for linearity
# The null hypothesis is that the model is linear.

# The linear_reset function can automatically create the higher-order terms
# from the fitted values by specifying the 'power' argument and 'test_type'.
# Setting power=3 will include the second and third powers of the fitted values.
reset_test = linear_reset(results, test_type='fitted', power=3)

print('\nRamsey RESET Test:')
print(reset_test)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_squared_error

# Read ZIP file
df = pd.read_csv('/content/HousePrices_HalfMil.csv.zip')

print("Original shape:", df.shape)

# Remove duplicate rows
df = df.drop_duplicates().copy()

print("After removing duplicates:", df.shape)

# Display columns
print("\nColumns:")
print(df.columns.tolist())

# Display important information
print("\nDataset information:")
print(df.info())

# Missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Select numeric columns
numeric_cols = df.select_dtypes(include=np.number).columns

print("Numeric columns:")
print(numeric_cols.tolist())

# Automatically select a likely price column
price_candidates = [
    col for col in numeric_cols
    if 'price' in col.lower()
]

if price_candidates:
    y_col = price_candidates[0]
else:
    y_col = numeric_cols[-1]

# Select first suitable numeric predictor different from price
x_candidates = [col for col in numeric_cols if col != y_col]

x_col = x_candidates[0]

print("\nIndependent variable:", x_col)
print("Dependent variable:", y_col)

# Remove missing values
data = df[[x_col, y_col]].dropna()

X = data[[x_col]]
y = data[y_col]

# Simple Linear Regression
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

# Residuals
residuals = y - y_pred

print("\nRegression Model:")
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nR² Score:", r2_score(y, y_pred))
print("Mean Squared Error:", mean_squared_error(y, y_pred))

print("\nFirst 10 Residuals:")
print(residuals.head(10))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_pred, residuals, alpha=0.4)

plt.axhline(y=0, linestyle='--')

plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.title("Residual Plot of Simple Linear Regression")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7, 7))

stats.probplot(residuals, dist="norm", plot=plt)

plt.title("Normal Probability Plot of Residuals")
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Load CSV file
df = pd.read_csv('/content/House Price Prediction Dataset.csv')

# Remove duplicate rows
df = df.drop_duplicates()

# Select Area and Price
data = df[['Area', 'Price']].dropna()

# Independent variable
X = data[['Area']]

# Dependent variable
y = data['Price']

# Create Simple Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Predicted values
fitted_values = model.predict(X)

# Residuals
residuals = y - fitted_values

print("Dataset shape:", df.shape)
print("Number of observations:", len(data))
print("Regression model created successfully.")
```

```python
# Regression coefficients
intercept = model.intercept_
slope = model.coef_[0]

# R² value
r2 = r2_score(y, fitted_values)

# Mean Squared Error
mse = mean_squared_error(y, fitted_values)

print("Simple Linear Regression Model")
print("--------------------------------")

print(f"Intercept = {intercept:.2f}")
print(f"Slope = {slope:.2f}")

print("\nRegression Equation:")
print(f"Price = {intercept:.2f} + ({slope:.2f} × Area)")

print(f"\nR² Value = {r2:.4f}")
print(f"Mean Squared Error = {mse:.2f}")

print("\nFirst 10 Residuals:")
print(residuals.head(10))
```
